```python
#Jelani Spencer-Joe

import json

# Define dictionary of movies and information
movie_database = {
    "Spider-Man": {
        "year": 2002,
        "genre": "Superhero",
        "director": "Sam Raimi",
        "actors": ["Tobey Maguire", "Kirsten Dunst", "Willem Dafoe"]
    },
    "Us": {
        "year": 2019,
        "genre": "Horror",
        "director": "Jordan Peele",
        "actors": ["Lupita Nyong'o", "Winston Duke", "Shahadi Wright Joseph"]
    },
    "High School Musical": {
        "year": 2006,
        "genre": "Musical",
        "director": "Kenny Ortega",
        "actors": ["Zac Efron", "Vanessa Hudgens", "Ashley Tisdale"]
    }
}


# Print out the movies and their information using square bracket syntax and .keys
for movie in movie_database.keys():
    print(f"Movie: {movie}")
    for key in movie_database[movie].keys():
        value = movie_database[movie][key]  # Square bracket syntax to get the value
        print(f"{key}: {value}")
    print()

# Update an element in the database
# Replace director in High School Musical to Steven Spielberg
movie_database["High School Musical"]["director"] = "Steven Spielberg"

# Delete an element in the database
# Delete Spider-Man from the database
    del movie_database["Spider-Man"]

# Get a value using .get()
genre = movie_database.get("Us", {}).get("genre")
print(f"The genre of Us is {genre}.\n")


# --- Functions for the Management System ---

def add_movie():
    title = input("Enter the movie title: ")
    genre = input("Enter the movie genre: ")
    director = input("Enter the name of the director: ")
    release_date = input("Enter the release date of the movie (YYYY-MM-DD): ")
    actors = input("Enter the names of the actors (separated by commas): ")
    actors_list = [actor.strip() for actor in actors.split(",")]

    movie_database[title] = {
        "genre": genre,
        "director": director,
        "release_date": release_date,
        "actors": actors_list
    }
    print(f"{title} has been added to the database.")

def edit_movie():
    title = input("Enter the movie title to edit: ")
    if title in movie_database:
        print(f"Current information for {title}:")
        print(json.dumps(movie_database[title], indent=4))

        genre = input("Enter the movie genre (or press Enter to keep current value): ")
        director = input("Enter the name of the director (or press Enter to keep current value): ")
        release_date = input("Enter the release date (YYYY-MM-DD) (or press Enter to keep current value): ")
        actors = input("Enter the names of the actors (separated by commas) (or press Enter to keep current value): ")

        if genre:
            movie_database[title]["genre"] = genre
        if director:
            movie_database[title]["director"] = director
        if release_date:
            movie_database[title]["release_date"] = release_date
        if actors:
            movie_database[title]["actors"] = [actor.strip() for actor in actors.split(",")]

        print(f"{title} has been updated.")
    else:
        print(f"{title} is not in the database.")

def delete_movie():
    title = input("Enter the movie title to delete: ")
    if title in movie_database:
        del movie_database[title]
        print(f"{title} has been deleted from the database.")
    else:
        print(f"{title} is not in the database.")

def view_all_movies():
    print("All movies in the database:")
    for movie, info in movie_database.items():
        print(f"{movie}: {json.dumps(info, indent=4)}")

def search_movies():
    print("Search movies in the database:")
    criteria = input("Enter the search criteria: ")
    matches = []
    for movie, info in movie_database.items():
        if (criteria.lower() in movie.lower() or
            criteria.lower() in info.get("genre", "").lower() or
            criteria.lower() in info.get("director", "").lower() or
            any(criteria.lower() in actor.lower() for actor in info.get("actors", ""))):
            matches.append(movie)

    if matches:
        print("Matches found:")
        for match in matches:
            print(f"{match}: {json.dumps(movie_database[match], indent=4)}")
    else:
     print("No matches found.")

def save_data():
    filename = input("Enter the filename to save to: ")
    with open(filename, "w") as file:
        json.dump(movie_database, file, indent=4)
    print("Data saved successfully.")

def load_data():
    global movie_database
    filename = input("Enter the filename to load from: ")
    try:
        with open(filename, "r") as file:
            movie_database = json.load(file)
        print("Data loaded successfully.")
    except FileNotFoundError:
        print("File not found.")


# --- Main Program Loop ---
if __name__ == "__main__":
    while True:
        print("\n==== Movie Database Management System ====\n")
        print("1. Add a new movie")
        print("2. Edit an existing movie")
        print("3. Delete a movie")
        print("4. View all movies")
        print("5. Search movies")
        print("6. Save data to file")
        print("7. Load data from file")
        print("8. Exit\n")

        choice = input("Enter your choice: ")

        if choice == "1":
            add_movie()
        elif choice == "2":
            edit_movie()
        elif choice == "3":
            delete_movie()
        elif choice == "4":
            view_all_movies()
        elif choice == "5":
            search_movies()
        elif choice == "6":
            save_data()
        elif choice == "7":
            load_data()
        elif choice == "8":
            print("Exiting program.")
            break
        else:
            print("Invalid choice. Please try again.")
```
